# Python Script to Jupyter Notebook Conversion
This notebook automates the conversion of Python scripts (`.py`) to Jupyter Notebooks (`.ipynb`), validates, executes, and exports them in VS Code.

## 1. Import and Verify Dependencies
Import required libraries and print their versions to confirm availability.

```python
import sys
import jupytext
import nbformat
import nbconvert
import nbclient

print('Python:', sys.version)
print('jupytext:', jupytext.__version__)
print('nbformat:', nbformat.__version__)
print('nbconvert:', nbconvert.__version__)
print('nbclient:', nbclient.__version__)
```

## 2. Install Missing Packages
Install `jupytext`, `nbformat`, `nbconvert`, and `nbclient` if they are not already available.

```python
import subprocess

def install(package):
    subprocess.check_call([sys.executable, "-m", "pip", "install", package])

for pkg in ["jupytext", "nbformat", "nbconvert", "nbclient"]:
    try:
        __import__(pkg)
    except ImportError:
        print(f"Installing {pkg}...")
        install(pkg)
```

## 3. Define Input/Output Paths
Set the source Python file and destination notebook file paths.

```python
import os
src_py = r"C:/Users/ASUS/Downloads/1_prepare_problem.py"
dst_ipynb = r"C:/Users/ASUS/Documents/SGU_khai_pha_du_lieu_DDU1231/Thực hành/New folder/EDA/1_prepare_problem.ipynb"
print(f"Source: {src_py}\nDestination: {dst_ipynb}")
```

## 4. Convert .py to .ipynb (Single File)
Use Jupytext to convert the Python script to a notebook.

```python
import jupytext
import nbformat

# Read .py and write .ipynb
nb = jupytext.read(src_py)
nbformat.write(nb, dst_ipynb)
print(f"Converted {src_py} to {dst_ipynb}")
```
